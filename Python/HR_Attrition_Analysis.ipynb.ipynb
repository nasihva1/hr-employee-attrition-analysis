### Exploratary Data Analysis

```python
# 1. Count employees by Overtime and Attrition
overtime_attrition = pd.crosstab(
    df_clean["OverTime"],
    df_clean["Attrition"]
)

# 2. Calculate attrition rate for each Overtime group
overtime_attrition["AttritionRate"] = (
    overtime_attrition["Yes"]
    / overtime_attrition.sum(axis=1)
) * 100

# 3. Display the result
overtime_attrition
```

```python
import matplotlib.pyplot as plt

# Create bar chart
plt.bar(
    overtime_attrition.index,
    overtime_attrition["AttritionRate"]
)

# Add labels
plt.xlabel("OverTime")
plt.ylabel("Attrition Rate (%)")
plt.title("Employee Attrition Rate by Overtime")

# Display the chart
plt.show()
```

```python
# Job Satisfaction vs Attrition

job_satisfaction_attrition = pd.crosstab(
    df_clean["JobSatisfaction"],
    df_clean["Attrition"]
)

job_satisfaction_attrition["AttritionRate"] = (
    job_satisfaction_attrition["Yes"]
    / job_satisfaction_attrition.sum(axis=1)
) * 100

print(job_satisfaction_attrition)

# Visualization
plt.figure(figsize=(8, 5))

plt.bar(
    job_satisfaction_attrition.index,
    job_satisfaction_attrition["AttritionRate"]
)

plt.xlabel("Job Satisfaction")
plt.ylabel("Attrition Rate (%)")
plt.title("Employee Attrition Rate by Job Satisfaction")
plt.xticks([1, 2, 3, 4])

plt.show()
```

```python
# Age vs Attrition

# 1. Create age groups
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[17, 25, 35, 45, 55, 60],
    labels=["18-25", "26-35", "36-45", "46-55", "56-60"]
)

# 2. Count Attrition by Age Group
age_attrition = pd.crosstab(
    df_clean["AgeGroup"],
    df_clean["Attrition"]
)

# 3. Calculate attrition rate
age_attrition["AttritionRate"] = (
    age_attrition["Yes"]
    / age_attrition.sum(axis=1)
) * 100

# 4. Display the result
print(age_attrition)

# 5. Visualize
plt.figure(figsize=(8, 5))

plt.bar(
    age_attrition.index,
    age_attrition["AttritionRate"]
)

plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.title("Employee Attrition Rate by Age Group")

plt.show()
```

```python
# Compare attrition rates across important categorical variables

variables = [
    "OverTime",
    "JobLevel",
    "JobRole",
    "BusinessTravel",
    "MaritalStatus",
    "Gender",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "JobInvolvement",
    "WorkLifeBalance",
    "RelationshipSatisfaction"
]

results = []

for col in variables:
    table = pd.crosstab(
        df_clean[col],
        df_clean["Attrition"],
        normalize="index"
    ) * 100

    for category in table.index:
        results.append({
            "Variable": col,
            "Category": category,
            "AttritionRate": table.loc[category, "Yes"]
        })

attrition_summary = pd.DataFrame(results)

attrition_summary = attrition_summary.sort_values(
    "AttritionRate",
    ascending=False
)

print(attrition_summary)
```

## Key Insights

- **Overall Attrition:** The overall employee attrition rate is **16.12%**, with **237 out of 1,470 employees** leaving the company.

- **Overtime:** Employees working overtime have a substantially higher attrition rate (**30.53%**) compared with employees not working overtime (**10.44%**).

- **Job Level:** **Job Level 1** has the highest attrition rate among job levels at **26.34%**, while higher job levels generally show lower attrition.

- **Job Role:** **Sales Representatives** have the highest attrition rate among job roles at **39.76%**, followed by Laboratory Technicians (**23.94%**) and Human Resources employees (**23.08%**).

- **Business Travel:** Employees who travel frequently have a higher attrition rate (**24.91%**) compared with employees who travel rarely (**14.96%**) or do not travel (**8.00%**).

- **Job Involvement:** Employees with the lowest job involvement rating (1) have a high attrition rate of **33.73%**, compared with **9.03%** for employees with the highest rating (4).

- **Work-Life Balance:** Employees with the lowest work-life balance rating (1) have a high attrition rate of **31.25%**.

- **Environment Satisfaction:** Employees with the lowest environment satisfaction rating (1) have an attrition rate of **25.35%**, compared with **13.45%** for rating 4.

- **Job Satisfaction:** Employees with the lowest job satisfaction rating (1) have an attrition rate of **22.84%**, compared with **11.33%** for rating 4.

- **Marital Status:** Single employees have a higher attrition rate (**25.53%**) than married (**12.48%**) or divorced employees (**10.09%**).

- **Department:** The Sales department has the highest departmental attrition rate (**20.63%**), compared with **13.84%** in Research & Development and **19.05%** in Human Resources.

- **Gender:** Male employees have a slightly higher attrition rate (**17.01%**) than female employees (**14.80%**), although the difference is relatively small compared with stronger factors.

- **Age:** Employees who left were younger on average (**33.61 years**) than employees who stayed (**37.56 years**).

- **Monthly Income:** Employees who left had a lower average monthly income (**4,787.09**) than employees who stayed (**6,832.74**).

- **Distance From Home:** Employees who left lived farther from work on average (**10.63**) than employees who stayed (**8.92**).

- **Work Experience:** Employees who left had fewer average total working years (**8.24**) than employees who stayed (**11.86**).

- **Company Tenure:** Employees who left had shorter average tenure at the company (**5.13 years**) than employees who stayed (**7.37 years**).

- **Current Role Tenure:** Employees who left had fewer average years in their current role (**2.90 years**) than employees who stayed (**4.48 years**).

- **Manager Tenure:** Employees who left had fewer average years with their current manager (**2.85 years**) than employees who stayed (**4.37 years**).

### Overall Conclusion

The strongest descriptive patterns are concentrated around **overtime, job level, job role, business travel, job involvement, work-life balance, satisfaction, and employee experience/tenure**.

These findings represent **associations rather than causal relationships**. Multiple factors may overlap with each other, so the strongest individual patterns should be investigated further using segmented analysis.
